```python
import pandas as pd
#creating the dataframes / importing the data
employees = pd.read_excel("../data_raw/Total_employees.xlsx")
unemployees = pd.read_excel("../data_raw/Unemployees.xlsx")
openings = pd.read_excel("../data_raw/Job_openings.xlsx")
layoffs = pd.read_excel("../data_raw/Layoff_rates.xlsx")
hires = pd.read_excel("../data_raw/Hire_rate.xlsx")


```

```python
#since it has a description I'm verifying by 15 rows if the data appears therefore it's reading it correctly
employees.head(15)
openings.head(15)
layoffs.head(15)
hires.head(15)
unemployees.head(15)
```

```python
unemployees.columns()
```

```python
#this line is to delete the description of the data that appears on the excel file.
employees = employees.iloc[11:]
openings = openings.iloc[12:]
layoffs = layoffs.iloc[12:]
hires = hires.iloc[12:]
unemployees = unemployees.iloc[10:]
```

```python
unemployees.head(15)
```

```python
#I'm resetting the index so it doesn't count the description rows that I deleted
employees = employees.reset_index(drop=True)
openings = openings.reset_index(drop=True)
layoffs = layoffs.reset_index(drop=True)
hires = hires.reset_index(drop=True)
unemployees = unemployees.reset_index(drop=True)
```

```python
#This is to verify if it has all the columns, in total should be 13 (12 due to the months and the extra for the title)
layoffs.info()
```

```python
unemployees.head()
```

```python
#This is to fix the error that the it was reading the wrong data and set it into the columns, so I redifined the right columns
employees.columns = employees.iloc[0]
employees = employees[1:].reset_index(drop=True)

openings.columns = openings.iloc[0]
openings = openings[1:].reset_index(drop=True)

layoffs.columns = layoffs.iloc[0]
layoffs = layoffs[1:].reset_index(drop=True)

hires.columns = hires.iloc[0]
hires = hires[1:].reset_index(drop=True)

unemployees.columns = unemployees.iloc[0]
unemployees = unemployees[1:].reset_index(drop=True)
```

```python
employees.head()
```

```python

#this creates a new column with the average from january to december(all months), then it groups it by year.
employees["Employees_Avg"] = employees.loc[:, "Jan":"Dec"].mean(axis=1)
employees_avg = employees[["Year", "Employees_Avg"]]

openings["Openings_Avg"] = openings.loc[:, "Jan":"Dec"].mean(axis=1)
openings_avg = openings[["Year", "Openings_Avg"]]
#Hires is a rate- type value
hires["Hires_Avg"] = hires.loc[:, "Jan":"Dec"].mean(axis=1)
hires_avg = hires[["Year", "Hires_Avg"]]
#layoffs is a rate-type value
layoffs["Layoffs_Avg"] = layoffs.loc[:, "Jan":"Dec"].mean(axis=1)
layoffs_avg = layoffs[["Year", "Layoffs_Avg"]]

unemployees["Unemployees_Avg"] = unemployees.loc[:, "Jan":"Dec"].mean(axis=1)
unemployees_avg = unemployees[["Year", "Unemployees_Avg"]]
```

```python
unemployees.head()
```

```python
#Now I merged properly grouping them by year. Also I set unemployees column next to employees for easier readability
merged = employees_avg.merge(unemployees_avg, on="Year") \
.merge(openings_avg, on="Year") \
.merge(hires_avg, on="Year") \
.merge(layoffs_avg, on="Year") 
```

```python
merged.head(15)
```

```python
#Converting them into real counts 
merged[["Employees_Avg","Unemployees_Avg","Openings_Avg"]] *=1000
merged.head(15)
```

```python
#analyzing the data
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#this starts a new plot and set the size of it
plt.figure(figsize=(16,6))
```

```python
sns.lineplot(data=merged, x="Year", y="Employees_Avg", label= "Employees")# I want to show a line graph showing how it has evolved during the years
plt.title("Number of Employees in the U.S") #Title
plt.xlabel("Year") #label for the x axis
plt.ylabel("Average")#label for y axis
plt.grid(True)#shows a grid behind the plot
plt.show()
```

```python
sns.lineplot(data=merged, x="Year", y="Unemployees_Avg", label= "Unemployees")# I want to show a line graph showing how it has evolved during the years
plt.show()
```

```python
#created a heatmap to show the correlation between variables
corr = merged.corr() 
plt.figure(figsize=(10,6))#size of the heatmap
sns.heatmap(corr, annot = True, cmap="coolwarm", fmt=".2f")#heatmap structure and showed by temperature
plt.title("Correlation map")
plt.show()
```

```python
sns.lineplot(data=merged, x="Year", y="Hires_Avg", label= "Hire rate vs Number of Employees")# I want to show a line graph showing how it has evolved during the years
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10,6))

# First variable (Hire rate)
ax1.plot(merged["Year"], merged["Hires_Avg"], color="tab:blue", label="Hire rate")
ax1.set_xlabel("Year")
ax1.set_ylabel("Hire rate", color="tab:blue")

# Second variable (Employees)
ax2 = ax1.twinx()
ax2.plot(merged["Year"], merged["Employees_Avg"], color="tab:red", label="Employees")
ax2.set_ylabel("Number of Employees", color="tab:red")

plt.title("Hire Rate impacted by Number of Employees (Dual Axis)")
fig.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10,6))
ax1.plot(merged["Year"], merged["Unemployees_Avg"], color="tab:green" , label="Unemployees")
ax1.set_xlabel("Year")
ax1.set_ylabel("Unemployees", color="tab:green")

ax2= ax1.twinx()
ax2.plot(merged["Year"], merged["Layoffs_Avg"], color="tab:orange", label="Layoff Rate")
ax2.set_ylabel("Layoff Rate", color="tab:orange")

plt.title("Correlation between Layoffs and Unemployment")
fig.tight_layout()
plt.show()
```
